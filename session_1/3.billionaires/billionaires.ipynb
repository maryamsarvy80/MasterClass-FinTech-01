# Compare Industries

```python
industry_counts = df["Industry"].value_counts()   
display(industry_counts.rename("Billionaires").to_frame())
```

# 7. Count billionaires in each industry

```python
# Top 10 industries stay separate
top_industries = industry_counts.head(10).copy()

top_industries["Other"] = industry_counts.iloc[10:].sum()

display(top_industries.rename("Billionaires").to_frame())
```

# 8.Create a pie chart

```python
colors = plt.cm.tab20.colors[:len(top_industries)]

fig, ax = plt.subplots(figsize=(10, 7))
wedges, _, autotexts = ax.pie(
    top_industries.values,
    autopct="%1.1f%%",
    startangle=90,
    counterclock=False,
    pctdistance=0.8,
    colors=colors,
    wedgeprops={"edgecolor": "white"}
)

legend_labels = [f"{name} ({count})" for name, count in top_industries.items()]
ax.legend(wedges, legend_labels, title="Industry (number of billionaires)",
          loc="center left", bbox_to_anchor=(1, 0.5))

ax.set_title("Billionaires by Industry (Top 10 + Other)")
plt.tight_layout()
plt.show()
```
